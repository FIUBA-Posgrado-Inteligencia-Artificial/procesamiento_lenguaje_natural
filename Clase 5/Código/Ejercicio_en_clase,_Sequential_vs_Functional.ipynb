# **Ejercicio resuelto, Sequential vs Functional**

Importamos lo necesario

```python
import numpy as np
import tensorflow as tf
from tensorflow.keras import layers, models, Input
```

Importamos los datos

```python
(X_train, y_train), (X_test, y_test) = tf.keras.datasets.mnist.load_data()
```

Aplanamos las imágenes 28x28 y normalizamos

```python
X_train = X_train.reshape(-1, 28 * 28).astype("float32") / 255.0
X_test = X_test.reshape(-1, 28 * 28).astype("float32") / 255.0

input_dim = X_train.shape[1]   # 784
num_clases = 10
```

### A) Modelo Sequential

```python
tf.keras.utils.set_random_seed(42)

modelo_seq = models.Sequential([
    layers.Dense(128, activation="relu", input_shape=(input_dim,), name="oculta_1"),
    layers.Dense(64, activation="relu", name="oculta_2"),
    layers.Dense(num_clases, activation="softmax", name="salida"),
], name="modelo_sequential")
```

Compilamos y entrenamos el modelo

```python
tf.keras.utils.set_random_seed(42)

modelo_seq.compile(optimizer="adam", loss="sparse_categorical_crossentropy", metrics=["accuracy"])
hist_seq = modelo_seq.fit(X_train, y_train, epochs=5, batch_size=32, verbose=1)

acc_seq = modelo_seq.evaluate(X_test, y_test, verbose=0)[1]
print(f"Accuracy en test (Sequential): {acc_seq:.4f}")
```

### B) Modelo Functional

```python
tf.keras.utils.set_random_seed(42)

entrada = Input(shape=(input_dim,), name="entrada")
x = layers.Dense(128, activation="relu", name="oculta_1")(entrada)
x = layers.Dense(64, activation="relu", name="oculta_2")(x)
salida = layers.Dense(num_clases, activation="softmax", name="salida")(x)

modelo_func = models.Model(inputs=entrada, outputs=salida, name="modelo_functional")

modelo_func.summary()
```

Compilamos y entrenamos el modelo

```python
tf.keras.utils.set_random_seed(42)

modelo_func.compile(optimizer="adam", loss="sparse_categorical_crossentropy", metrics=["accuracy"])
hist_func = modelo_func.fit(X_train, y_train, epochs=5, batch_size=32, verbose=1)

acc_func = modelo_func.evaluate(X_test, y_test, verbose=0)[1]
print(f"Accuracy en test (Functional): {acc_func:.4f}")
```
